```python
import osmnx
from optalgotools.structures import Node
from optalgotools.routing import cost, draw_route
from optalgotools.algorithms.graph_search import (
    BFS,
    DFS,
    Dijkstra,
    UCS,
    Bidirectional_Dijkstra,
)

# King's College Circle, Toronto
reference = (43.661667, -79.395)

G = osmnx.graph_from_point(
    reference,
    dist=300,
    network_type="walk",
    simplify=True,
)

# Route endpoints
origin_point = (43.664527, -79.392442) # King Edward VII Equestrian Statue
destination_point = (43.659659, -79.397669) # Bahen Centre for Information Technology at UofT

# Find the nearest graph nodes.
# nearest_nodes expects X=longitude, Y=latitude.
origin_id = osmnx.distance.nearest_nodes(
    G,
    X=origin_point[1],
    Y=origin_point[0],
)

destination_id = osmnx.distance.nearest_nodes(
    G,
    X=destination_point[1],
    Y=destination_point[0],
)

# Convert them into optalgotools Node objects
origin = Node(graph=G, osmid=origin_id)
destination = Node(graph=G, osmid=destination_id)

print("Origin node:", origin_id)
print("Destination node:", destination_id)
```

```python
solution = BFS(origin, destination)
route = solution.result
print(f"Cost: {cost(G,route)} m")
print(f"Process time: {solution.time} s")
print(f"Space required: {solution.space} bytes")
print(f"Explored nodes: {solution.explored}")
draw_route(G,route)
```

```python
solution = DFS(origin, destination)
route = solution.result
print(f"Cost: {cost(G,route)} m")
print(f"Process time: {solution.time} s")
print(f"Space required: {solution.space} bytes")
print(f"Explored nodes: {solution.explored}")
draw_route(G,route)
```

```python
unrelaxed_nodes = [Node(G, osmid) for osmid in G.nodes()]
solution = Dijkstra(origin, destination, unrelaxed_nodes)
route = solution.result
print(f"Cost: {cost(G,route)} m")
print(f"Process time: {solution.time} s")
print(f"Space required: {solution.space} bytes")
print(f"Explored nodes: {solution.explored}")
draw_route(G,route)
```

```python
solution = UCS(origin, destination)
route = solution.result
print(f"Cost: {cost(G,route)} m")
print(f"Process time: {solution.time} s")
print(f"Space required: {solution.space} bytes")
print(f"Explored nodes: {solution.explored}")
draw_route(G,route)
```

```python
unrelaxed_nodes = [Node(G,osmid) for osmid in G.nodes()]
solution = Bidirectional_Dijkstra(origin, destination, unrelaxed_nodes)
route = solution.result
print(f"Cost: {cost(G,route)} m")
print(f"Process time: {solution.time} s")
print(f"Space required: {solution.space} bytes")
print(f"Explored nodes: {solution.explored}")
draw_route(G,route)
```
